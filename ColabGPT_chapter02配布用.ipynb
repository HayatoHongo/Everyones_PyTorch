4つの異なる学習率 (`0.1`, `0.3`, `0.001`, `1`) での勾配降下法の結果を比較すると、以下の傾向が見られます。

*   **`lr = 0.1` (元の設定):** 最も速く最適な `w` (約2.0) と最小の損失に収束しています。<br>
*   **`lr = 0.3`:** 重みwは2の左右を行ったり来たりしながらも徐々に目標値に近づいています。<br>

後半の2つでは学習率の桁が変わります。
*   **`lr = 0.001`:** さらに収束が遅く、10エポックではまだ目標値から遠いことがわかります。<br>
*   **`lr = 1`:** 今度は重みwの振れ幅が大きすぎて、$w=2$になる気配がありません。これを **勾配の発散** と言います。<br>

学習率が小さすぎると学習に時間がかかり、逆に大きすぎると最適な値を行き過ぎて振動したり、発散したりする可能性があります。<br>
**適切な学習率を選択すること** は効率的な学習において非常に重要なのです。

  ---

  # 番外編① 学習率をいじるとどうなる？
  本章で用いた損失関数と重みの更新式を振り返ります。

  *  損失関数: $L = 2.5w^2-10w+10$
  * 勾配(grad): $L' = 5w -10$
  * 重みの更新式: $w ← w - lr × grad$

  また、重み$w$の初期値は0に設定しました。<br>
  学習率$lr$を調整することで、重みwの挙動は大きく変化します。
  これまで見てきた学習率($lr$ = `0.1`, `0.3`, `0.001`, `1`)では、$w$はループすることなく、毎回違う値に更新されました。実は$lr$をある値にすると、重み$w$が同じ値をループします。<br>

  >**問題**<br>
  >重みwの更新式を$w_{t+1}$ ← $w_{t} - lr × grad$とする。このとき、$w_{t+2} = w_{t}$となる学習率$lr$を求めよ。<br>
  ただし$lr > 0$とし、損失関数は$L = 2.5w^2-10w+10$を用いる。また、$w_{0}=0$とする。


手計算で求めるのも良し、下のコードで$lr$をいじって求めるのも良し。

````python
w = 0.0  # 再度リセット
lr = 0.1

for epoch in range(10):
    # 勾配の計算
    grad = 5 * w - 10

    # 更新
    w = w - lr * grad

    # 現在の損失 L = 2.5 * w^2 - 10 * w + 10 を計算
    loss = 2.5 * w**2 - 10 * w + 10

    print(f"エポック {epoch+1}: w = {w:.3f}, 損失 = {loss:.6f}")
````

````python
#@title 重み更新シュミレーター
from IPython.display import HTML

# コンパクト＆ダークモード版：L = 2.5 * (w - 2)^2 シミュレーター
html_code = """
<div id="sim-container" style="max-width: 450px; padding: 12px; border: 1px solid #444; border-radius: 8px; font-family: sans-serif; background-color: #1e1e1e; color: #f8f9fa;">
    <h4 style="margin-top:0; margin-bottom: 8px; color: #ffffff;">最急降下法の挙動シミュレーター</h4>
    <p style="font-size: 0.85em; margin: 0 0 10px 0; color: #adb5bd;">
        式: L = 2.5 * (w - 2)^2 &nbsp;|&nbsp; 初期値: w = 0
    </p>
    <div style="margin-bottom: 10px;">
        <label style="font-size: 0.9em;">学習率 (lr): <span id="lr-val" style="font-weight:bold; color:#4dabf7;">0.10</span></label><br>
        <input type="range" id="lr-slider" min="0.01" max="0.5" step="0.01" value="0.1" style="width: 100%; margin-top: 5px;">
    </div>
    <canvas id="canvas" width="420" height="200" style="background: #121212; width: 100%; border: 1px solid #333; border-radius: 4px; display: block;"></canvas>
    <div id="status" style="margin-top: 8px; font-weight: bold; font-size: 0.95em;"></div>
    <div id="logs" style="font-family: monospace; font-size: 0.75em; margin-top: 8px; color: #ced4da; height: 50px; overflow-y: auto; border-top: 1px solid #444; padding-top: 8px;"></div>
</div>

<script>
(function() {
    function init() {
        const canvas = document.getElementById('canvas');
        if (!canvas || !canvas.getContext) return;
        const ctx = canvas.getContext('2d');
        const slider = document.getElementById('lr-slider');
        const lrVal = document.getElementById('lr-val');
        const status = document.getElementById('status');
        const logs = document.getElementById('logs');

        function draw() {
            const lr = parseFloat(slider.value);
            lrVal.innerText = lr.toFixed(2);

            ctx.clearRect(0, 0, canvas.width, canvas.height);

            // キャンバスが小さくなった分、余白(padding)も小さく調整
            const padding = 35;
            const width = canvas.width - padding * 2;
            const height = canvas.height - padding * 2;

            const w_min = -1.5;
            const w_max = 5.5;
            const l_max = 20;

            const w2x = (w) => padding + ((w - w_min) / (w_max - w_min)) * width;
            const l2y = (l) => (canvas.height - padding) - (l / l_max) * height;

            // 1. 損失関数の描画
            ctx.beginPath();
            ctx.strokeStyle = '#868e96';
            ctx.lineWidth = 1.5;
            for(let w = w_min; w <= w_max; w += 0.05) {
                let l = 2.5 * Math.pow(w - 2, 2);
                if(l > l_max) continue;

                let lx = w2x(w);
                let ly = l2y(l);

                if (ly < padding || ly > (canvas.height - padding)) continue;

                if (w === w_min || 2.5 * Math.pow(w - 0.05 - 2, 2) > l_max) ctx.moveTo(lx, ly);
                else ctx.lineTo(lx, ly);
            }
            ctx.stroke();

            // 2. 最急降下法の計算
            let w = 0;
            let path = [w];
            for(let i=0; i<15; i++) {
                let grad = 5 * (w - 2);
                w = w - lr * grad;
                path.push(w);
                if (w < w_min * 2 || w > w_max * 2) break;
            }

            // 3. 軌跡の白い線
            ctx.beginPath();
            ctx.strokeStyle = '#ffffff';
            ctx.lineWidth = 1.5;
            let isFirstLinePoint = true;

            path.forEach((val) => {
                let loss = 2.5 * Math.pow(val - 2, 2);
                let lx = w2x(val);
                let ly = l2y(loss);

                if (lx >= padding && lx <= (canvas.width - padding) && ly >= padding && ly <= (canvas.height - padding)) {
                    if(isFirstLinePoint) {
                        ctx.moveTo(lx, ly);
                        isFirstLinePoint = false;
                    } else {
                        ctx.lineTo(lx, ly);
                    }
                }
            });
            ctx.stroke();

            // 4. 点の描画
            path.forEach((val, i) => {
                let loss = 2.5 * Math.pow(val - 2, 2);
                let lx = w2x(val);
                let ly = l2y(loss);

                if (lx >= padding && lx <= (canvas.width - padding) && ly >= padding && ly <= (canvas.height - padding)) {
                    ctx.fillStyle = i === 0 ? '#fcc419' : '#339af0';
                    ctx.beginPath();
                    // キャンバスが小さいため、点のサイズも少し小さく(4.5 -> 3.5)
                    ctx.arc(lx, ly, 3.5, 0, Math.PI*2);
                    ctx.fill();
                }
            });

            // 5. 状態判定
            if (lr < 0.4) {
                status.innerText = "状態: 収束 (w=2 に接近)";
                status.style.color = "#51cf66";
            } else if (Math.abs(lr - 0.4) < 0.0001) {
                status.innerText = "状態: 振動 (w=0 と w=4 を反復)";
                status.style.color = "#fcc419";
            } else {
                status.innerText = "状態: 発散 (歩幅が大きすぎます)";
                status.style.color = "#ff6b6b";
            }

            logs.innerHTML = path.slice(0, 5).map((v, i) => `Step ${i}: w = ${v.toFixed(3)}`).join('<br>') + (path.length > 5 ? '<br>...' : '');
        }

        slider.addEventListener('input', draw);
        draw();
    }

    if (document.readyState === 'complete' || document.readyState === 'interactive') {
        init();
    } else {
        document.addEventListener('DOMContentLoaded', init);
    }
})();
</script>
"""

display(HTML(html_code))
````
